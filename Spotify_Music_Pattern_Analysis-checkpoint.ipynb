```python
#Dataset load karo
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Spotify dataset
file_path = r"C:\Users\nishika tendane\Downloads\archive (7)\dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python

print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
#Missing values aur duplicate rows check karo

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nDataset shape:", df.shape)
```

```python
#Data cleaning karo

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Fill missing numeric values with median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with mode, if available
categorical_cols = df.select_dtypes(
    include=["object", "category", "string"]
).columns

for col in categorical_cols:
    if df[col].isnull().any():
        mode_values = df[col].mode()
        if not mode_values.empty:
            df[col] = df[col].fillna(mode_values.iloc[0])

print("Cleaning completed!")
print("Remaining missing values:", df.isnull().sum().sum())
print("New dataset shape:", df.shape)
```

```python
#Column names check karo
print(df.columns.tolist())
```

```python
#Most popular songs find karo

if "popularity" in df.columns:
    sort_cols = ["popularity"]
    ascending_order = [False]

    if "track_name" in df.columns:
        sort_cols.append("track_name")
        ascending_order.append(True)

    top_songs = df.sort_values(
        sort_cols, ascending=ascending_order
    ).head(10)

    display(top_songs)
else:
    print("Popularity column nahi mila. Step 6 ke column names check karo.")
```

```python
#Danceability aur energy ka analysis

features = [
    col for col in ["danceability", "energy", "valence", "acousticness"]
    if col in df.columns
]

if features:
    df[features].hist(figsize=(12, 8), bins=25)
    plt.suptitle("Spotify Music Audio Feature Distributions")
    plt.tight_layout()
    plt.show()
else:
    print("Expected audio feature columns nahi mile.")
```

```python
#Correlation heatmap banao

audio_features = [
    col for col in [
        "popularity", "danceability", "energy",
        "valence", "tempo", "acousticness",
        "instrumentalness", "loudness", "speechiness"
    ]
    if col in df.columns and pd.api.types.is_numeric_dtype(df[col])
]

if len(audio_features) >= 2:
    corr = df[audio_features].corr()

    plt.figure(figsize=(11, 7))
    sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Spotify Audio Features Correlation")
    plt.tight_layout()
    plt.show()
else:
    print("Correlation ke liye enough numeric columns nahi mile.")
```

```python
#Final summary aur cleaned CSV save karo

print("SPOTIFY MUSIC PATTERN ANALYSIS")
print("-" * 40)

print("Total tracks:", len(df))
print("Total columns:", len(df.columns))

if "popularity" in df.columns:
    print("Average popularity:", round(df["popularity"].mean(), 2))

if "danceability" in df.columns:
    print("Average danceability:", round(df["danceability"].mean(), 3))

if "energy" in df.columns:
    print("Average energy:", round(df["energy"].mean(), 3))

output_path = (
    r"C:\Users\nishika tendane\Downloads"
    r"\Spotify_Cleaned_Dataset.csv"
)

df.to_csv(output_path, index=False)
print("\nCleaned dataset saved to:", output_path)
```

```python
#Top 10 popular songs ka bar chart

if "popularity" in df.columns:
    top10 = df.nlargest(10, "popularity").copy()

    name_col = "track_name" if "track_name" in df.columns else None
    if name_col:
        labels = top10[name_col].astype(str)
    else:
        labels = top10.index.astype(str)

    plt.figure(figsize=(12, 6))
    plt.barh(labels[::-1], top10["popularity"].iloc[::-1])
    plt.xlabel("Popularity")
    plt.ylabel("Track")
    plt.title("Top 10 Most Popular Spotify Tracks")
    plt.tight_layout()
    plt.show()
else:
    print("Popularity column not found.")
```

```python
#Danceability vs Popularity

if {"danceability", "popularity"}.issubset(df.columns):
    sample = df[["danceability", "popularity"]].dropna()

    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=sample, x="danceability", y="popularity", alpha=0.4)
    plt.title("Danceability vs Popularity")
    plt.tight_layout()
    plt.show()

    print("Correlation:",
          round(sample["danceability"].corr(sample["popularity"]), 3))
else:
    print("Required columns not found.")
```

```python
#Energy vs Popularity

if {"energy", "popularity"}.issubset(df.columns):
    plt.figure(figsize=(8, 5))
    sns.regplot(
        data=df, x="energy", y="popularity",
        scatter_kws={"alpha": 0.25, "s": 15},
        line_kws={"color": "red"}
    )
    plt.title("Energy vs Popularity")
    plt.tight_layout()
    plt.show()
else:
    print("Required columns not found.")
```

```python
#Artists with the most tracks

artist_col = next(
    (c for c in ["artists", "artist_name", "artist"] if c in df.columns),
    None
)

if artist_col:
    artist_counts = df[artist_col].value_counts().head(10)

    plt.figure(figsize=(10, 6))
    artist_counts.sort_values().plot(kind="barh")
    plt.title("Top 10 Artists by Track Count")
    plt.xlabel("Number of tracks")
    plt.tight_layout()
    plt.show()

    display(artist_counts.rename("track_count").to_frame())
else:
    print("Artist column not found. Check df.columns.tolist().")
```

```python
#Audio features ka average

audio_cols = [
    c for c in [
        "danceability", "energy", "valence",
        "acousticness", "instrumentalness", "speechiness"
    ]
    if c in df.columns
]

if audio_cols:
    averages = df[audio_cols].mean().sort_values(ascending=False)
    display(averages.rename("average_value").to_frame())

    averages.sort_values().plot(kind="barh", figsize=(9, 5))
    plt.title("Average Spotify Audio Features")
    plt.xlabel("Average value")
    plt.tight_layout()
    plt.show()
else:
    print("Audio feature columns not found.")
```

```python
#Explicit songs analysis

if "explicit" in df.columns:
    explicit_counts = df["explicit"].value_counts(dropna=False)
    display(explicit_counts.rename("track_count").to_frame())

    explicit_counts.plot(kind="bar", figsize=(6, 4))
    plt.title("Explicit vs Non-explicit Tracks")
    plt.xlabel("Explicit")
    plt.ylabel("Number of tracks")
    plt.tight_layout()
    plt.show()
else:
    print("Explicit column not found.")
```

```python
#Release year trend

year_col = next(
    (c for c in ["year", "release_year"] if c in df.columns),
    None
)

if year_col:
    yearly_counts = df[year_col].value_counts().sort_index()
    yearly_counts = yearly_counts[yearly_counts.index.notna()]

    plt.figure(figsize=(12, 5))
    plt.plot(yearly_counts.index, yearly_counts.values)
    plt.title("Tracks by Release Year")
    plt.xlabel("Release year")
    plt.ylabel("Number of tracks")
    plt.tight_layout()
    plt.show()
else:
    print("Year column not found. If you have a release_date column, tell me.")
```

```python
#Genre analysis

genre_col = next(
    (c for c in ["track_genre", "genre", "playlist_genre"] if c in df.columns),
    None
)

if genre_col:
    genre_summary = (
        df.groupby(genre_col)
        .agg(
            track_count=(genre_col, "size"),
            average_popularity=("popularity", "mean")
        )
        .sort_values("average_popularity", ascending=False)
        .head(10)
        if "popularity" in df.columns
        else df[genre_col].value_counts().head(10).to_frame("track_count")
    )
    display(genre_summary)
else:
    print("Genre column not found in this dataset.")
```

```python
#Final key findings generate karo

print("SPOTIFY MUSIC PATTERN ANALYSIS: KEY FINDINGS")
print("=" * 50)
print(f"Total records: {len(df):,}")
print(f"Total columns: {len(df.columns)}")

for col in ["popularity", "danceability", "energy", "valence", "tempo"]:
    if col in df.columns:
        print(f"Average {col}: {df[col].mean():.3f}")

if {"danceability", "popularity"}.issubset(df.columns):
    print(
        "Danceability-popularity correlation:",
        round(df["danceability"].corr(df["popularity"]), 3)
    )

if {"energy", "popularity"}.issubset(df.columns):
    print(
        "Energy-popularity correlation:",
        round(df["energy"].corr(df["popularity"]), 3)
    )
```

```python
#Cleaned dataset aur notebook save karo

output_path = (
    r"C:\Users\nishika tendane\Downloads"
    r"\Spotify_Cleaned_Dataset.csv"
)

df.to_csv(output_path, index=False)
print("Saved:", output_path)
```
